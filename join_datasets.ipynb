```python
consumer_price_df = pd.read_csv("Consumer Prices Food Indices 2013.csv")
food_insecurity_df = pd.read_csv("Food Insecurity Indicator.csv")
food_inflation_df = pd.read_csv("Food Price Inflation FAO 23014.csv")
```

```python
consumer_price_df["year"] = consumer_price_df["TIME_PERIOD"].astype(str).str[:4].astype(int)
food_inflation_df["year"] = food_inflation_df["TIME_PERIOD"].astype(str).str[:4].astype(int)
```

```python
annual_consumer_price = (
    consumer_price_df.groupby(["REF_AREA", "year"])["OBS_VALUE"]
    .mean()
    .reset_index()
    .rename(columns={"OBS_VALUE": "annual_consumer_price"})
)
```

```python
annual_inflation = (
    food_inflation_df.groupby(["REF_AREA", "year"])["OBS_VALUE"]
    .mean()
    .reset_index()
    .rename(columns={"OBS_VALUE": "annual_food_inflation"})
)
```

```python
def calc_3_yr_avg(df, value_col):
    df = df.sort_values(["REF_AREA", "year"]).copy()
    new_col = value_col + "_3yr"
    df[new_col] = float("nan")

    for country, group in df.groupby("REF_AREA"):
        yearly = group.set_index("year")[value_col]

        all_years = range(yearly.index.min(), yearly.index.max() + 1)
        yearly = yearly.reindex(all_years)

        rolling_avg = yearly.rolling(window=3, min_periods=3).mean()

        df.loc[group.index, new_col] = (
            rolling_avg.reindex(group["year"]).values
        )

    return df
```

```python
annual_consumer_price = calc_3_yr_avg(
    annual_consumer_price, "annual_consumer_price"
)

annual_inflation = calc_3_yr_avg(
    annual_inflation, "annual_food_inflation"
)
```

```python
print(annual_consumer_price.columns)
print(annual_inflation.columns)
```

```python
food_insecurity_df["year"] = food_insecurity_df["TIME_PERIOD"].astype(int)
```

```python
food_insecurity_df = food_insecurity_df[
    food_insecurity_df["SEX_LABEL"] == "Total"
].copy()

food_insecurity_df["year"] = food_insecurity_df["TIME_PERIOD"].astype(int)

annual_food_insecurity = food_insecurity_df[
    ["REF_AREA", "year", "OBS_VALUE"]
].rename(columns={"OBS_VALUE": "annual_food_insecurity"})
```

```python
merged_df = annual_consumer_price[
    ["REF_AREA", "year", "annual_consumer_price_3yr"]
].merge(
    annual_inflation[
        ["REF_AREA", "year", "annual_food_inflation_3yr"]
    ],
    on=["REF_AREA", "year"],
    how="outer"
)

merged_df = merged_df.merge(
    annual_food_insecurity,
    on=["REF_AREA", "year"],
    how="outer"
)

merged_df = merged_df.sort_values(
    ["REF_AREA", "year"]
).reset_index(drop=True)
```

```python
merged_df.to_csv("Annual Food Datasets Merged.csv", index=False)
```
