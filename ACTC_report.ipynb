### Accuracy

checking extremities

```python
print('FS rows by SEX:', fs.SEX.value_counts().to_dict())
fs_tot = fs[fs.SEX == '_T'].copy()           # keep total-population rows only
print('FS total rows:', len(fs_tot), '| duplicates now:', fs_tot.duplicated(['REF_AREA', 'TIME_PERIOD']).sum(), '| countries:', fs_tot.REF_AREA.nunique())

print('\nLargest food CPI index values:')
print(idx.nlargest(5, 'OBS_VALUE')[['REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']])
print('Areas with index > 1000:', idx[idx.OBS_VALUE > 1000].REF_AREA_LABEL.unique().tolist())

print('\nLargest food inflation values (%):')
print(inf.nlargest(8, 'OBS_VALUE')[['REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']])
print('Observations with inflation > 100%:', (inf.OBS_VALUE > 100).sum())
print('Areas:', inf[inf.OBS_VALUE > 100].REF_AREA_LABEL.unique().tolist())
print('\nMost negative inflation values:')
print(inf.nsmallest(3, 'OBS_VALUE')[['REF_AREA_LABEL', 'TIME_PERIOD', 'OBS_VALUE']])
print('\nAreas with missing values:', inf[inf.OBS_VALUE.isna()].REF_AREA.value_counts().to_dict())
```

### Completeness:
checking for missing areas and timestamps

```python
#Time coverage per area in the price files
for n, df in [('index (23013)', idx), ('inflation (23014)', inf)]:
    g = df.groupby('REF_AREA').TIME_PERIOD.agg(['min', 'max', 'count'])
    print(n, '| areas starting after 2010:', (g['min'] > '2010').sum(),
          '| areas ending before 2025-01:', (g['max'] < '2025-01').sum(),
          '| median months per area:', g['count'].median(), '| min:', g['count'].min())
    print('  areas ending before 2025:', g[g['max'] < '2025-01'].index.tolist())
    print('  latest-month distribution:', g['max'].value_counts().head(3).to_dict())

#Missing months inside each area's own date range
def gap_report(df):
    d = df.copy(); d['p'] = pd.PeriodIndex(d.TIME_PERIOD, freq='M')
    out = {}
    for a, g in d.groupby('REF_AREA'):
        out[a] = ((g.p.max() - g.p.min()).n + 1) - len(g)
    s = pd.Series(out)
    return (s > 0).sum(), s[s > 0].to_dict()
print('\nIndex gaps (areas, months):', gap_report(idx))
print('Inflation gaps (areas, months):', gap_report(inf))
```

```python
#Food insecurity file coverage
print('FS countries per window end-year:', fs_tot.groupby('TIME_PERIOD').REF_AREA.nunique().to_dict())
print(fs_tot.groupby('TIME_PERIOD').OBS_VALUE.agg(['sum', 'median', 'max']).round(1))

big = ['IND', 'CHN', 'BRA', 'NGA', 'PAK', 'IDN', 'BGD', 'ETH', 'COD', 'USA', 'SDN', 'SOM', 'SYR', 'VEN', 'AFG', 'IRQ']
print('\nLarge/high-need countries absent from FS file:', [c for c in big if c not in set(fs_tot.REF_AREA)])
print('Price-file areas not in FS file:', len(set(inf.REF_AREA) - set(fs_tot.REF_AREA)))
print('FS countries absent from price files:', set(fs_tot.REF_AREA) - set(inf.REF_AREA))
print('\nWindow definitions (COMMENT_OBS):')
print(fs.COMMENT_OBS.value_counts().sort_index())
```

### Consistency

```python
i = idx.dropna(subset=['OBS_VALUE']).copy()
i['p'] = pd.PeriodIndex(i.TIME_PERIOD, freq='M')
piv = i.pivot(index='p', columns='REF_AREA', values='OBS_VALUE')
yoy = (piv / piv.shift(12) - 1) * 100
calc = yoy.stack().rename('calc').reset_index()

f = inf.dropna(subset=['OBS_VALUE']).copy()
f['p'] = pd.PeriodIndex(f.TIME_PERIOD, freq='M')
m = f.merge(calc, on=['REF_AREA', 'p'])
m['abs_diff'] = (m.OBS_VALUE - m.calc).abs()

print('Observations compared:', len(m))
print('Median abs diff (pp):', m.abs_diff.median().round(4))
print('Share within 0.1 pp:', (m.abs_diff < 0.1).mean().round(3), '| within 1 pp:', (m.abs_diff < 1).mean().round(3))
print('Inflation rows with no index counterpart:', len(f) - len(m))
print(m.nlargest(5, 'abs_diff')[['REF_AREA', 'p', 'OBS_VALUE', 'calc']])
```
